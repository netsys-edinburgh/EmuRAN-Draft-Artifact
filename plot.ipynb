# Reproduce and display the Chronos figures

Each code cell independently runs one figure's plotting script from `results/data`, writes a fresh PDF to `results/output`, and displays that generated PDF inline. Figure labels correspond to the draft.


```python
import os
import shutil
import subprocess
import sys
from pathlib import Path
from IPython.display import Image, Markdown, display

PROJECT_DIR = Path.cwd().resolve()
if not (PROJECT_DIR / "results").is_dir():
    raise RuntimeError(
        "Run this notebook from the project root "
        "(the directory containing the results folder)."
    )
RESULTS_DIR = PROJECT_DIR / "results"
DATA_DIR = RESULTS_DIR / "data"
OUTPUT_DIR = RESULTS_DIR / "output"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
import pymupdf

figure_dir = DATA_DIR / 'fig2'
plot_script = figure_dir / 'gen_latencies_cdf.py'
output_pdf = OUTPUT_DIR / 'fig2.pdf'
output_pdf.unlink(missing_ok=True)
command = [sys.executable, str(plot_script)] + []
environment = os.environ.copy()
environment["MPLBACKEND"] = "Agg"
display(Markdown('## Draft Figure 2 — Compute interruption and network-latency CDFs across cloud providers' + f"\n\nRunning `{plot_script.relative_to(PROJECT_DIR)}`…"))
completed = subprocess.run(
    command,
    cwd=figure_dir,
    env=environment,
    text=True,
    capture_output=True,
)
if completed.returncode != 0:
    raise RuntimeError(
        f"Plot script failed with exit code {completed.returncode}:\n"
        f"{completed.stderr.strip() or completed.stdout.strip()}"
    )
if completed.stdout.strip():
    print(completed.stdout.strip())
produced_pdf = figure_dir / 'fig2.pdf'
if not produced_pdf.exists():
    raise FileNotFoundError(f"Plot script did not generate {produced_pdf}")
shutil.move(str(produced_pdf), str(output_pdf))
if not output_pdf.exists():
    raise FileNotFoundError(f"Expected generated PDF at {output_pdf}")

document = pymupdf.open(output_pdf)
for page in document:
    pixmap = page.get_pixmap(matrix=pymupdf.Matrix(2, 2), alpha=False)
    display(Image(data=pixmap.tobytes("png")))
document.close()
display(Markdown(f"Generated: `{output_pdf.relative_to(PROJECT_DIR)}`"))
```
